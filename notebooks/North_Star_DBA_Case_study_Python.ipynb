III. Data Cleaning and Integration

### Cleaning and Integration Objective

This section applies the preparation steps identified during EDA. The workflow standardises zone values, fills selected missing values, prevents duplicate order rows when complaints are joined, verifies merge coverage, and creates delivery outcome fields for analysis.

Step 1: Standardise Zone Names



```python
# Standardise zone names before grouping, joining, or comparing zone-level performance.
# Raw files contain values such as AIRPORT, CENTRAL, Ctr, SOUTH, and mixed-case variants.
def clean_zone(series):
    return series.str.lower().str.strip().map({
        'north': 'North', 'south': 'South', 'east': 'East', 'west': 'West',
        'central': 'Central', 'ctr': 'Central', 'airport': 'Airport',
        'riverside': 'Riverside', 'riverside': 'Riverside'
    }).fillna(series)

customers['home_zone'] = clean_zone(customers['home_zone'])
orders['pickup_zone'] = clean_zone(orders['pickup_zone'])
orders['dropoff_zone'] = clean_zone(orders['dropoff_zone'])
drivers['base_zone'] = clean_zone(drivers['base_zone'])
vehicles['assigned_zone'] = clean_zone(vehicles['assigned_zone'])
app_events['zone_context'] = clean_zone(app_events['zone_context'])
# Verify standardised zone labels before continuing with analysis.
print("Customers zones:", sorted(customers['home_zone'].dropna().unique()))
print("Orders pickup zones:", sorted(orders['pickup_zone'].dropna().unique()))
print("Orders dropoff zones:", sorted(orders['dropoff_zone'].dropna().unique()))
print("Drivers base zones:", sorted(drivers['base_zone'].dropna().unique()))
print("Vehicles assigned zones:", sorted(vehicles['assigned_zone'].dropna().unique()))
print("App events zones:", sorted(app_events['zone_context'].dropna().unique()))
```

Step 2: Handle Missing Values



```python
# Fill missing values based on EDA findings

# Customers - loyalty scores and preferred channel
customers['loyalty_score'] = customers['loyalty_score'].fillna(customers['loyalty_score'].median())
customers['preferred_channel'] = customers['preferred_channel'].fillna('Unknown')

# Orders - booking channel
orders['booking_channel'] = orders['booking_channel'].fillna('Unknown')

# Deliveries - customer ratings (0 = not rated)
deliveries['customer_rating_post_delivery'] = deliveries['customer_rating_post_delivery'].fillna(0)

# Drivers - training scores
drivers['training_score'] = drivers['training_score'].fillna(drivers['training_score'].median())

# Vehicles - battery health
vehicles['battery_health_pct'] = vehicles['battery_health_pct'].fillna(vehicles['battery_health_pct'].median())

# Complaints - compensation amount (0 = no payout)
complaints['compensation_amount'] = complaints['compensation_amount'].fillna(0)

# App_events - no fill needed, 144 missing order_ids are expected

# Verify
print("Customers missing:", customers.isna().sum().sum())
print("Orders missing:", orders.isna().sum().sum())
print("Deliveries missing:", deliveries.isna().sum().sum())
print("Drivers missing:", drivers.isna().sum().sum())
print("Vehicles missing:", vehicles.isna().sum().sum())
print("Complaints missing:", complaints.isna().sum().sum())
```

Step 3: Merge into a Single Consolidated Dataframe

```python
# Check row counts after each join to identify duplication
print("Orders:", len(orders))
print("+ Deliveries:", len(orders.merge(deliveries, on='order_id', how='left')))
print("+ Customers:", len(orders.merge(deliveries, on='order_id', how='left').merge(customers, on='customer_id', how='left')))
print("+ Complaints:", len(orders.merge(deliveries, on='order_id', how='left').merge(customers, on='customer_id', how='left').merge(complaints, on='order_id', how='left', suffixes=('', '_complaint'))))
```

The row count remained at 1,250 until the complaints table was joined, where it increased to 1,285. This confirms that 35 orders have more than one complaint record, causing duplication.
To prevent duplication, only the first complaint per order was kept. This ensures each order appears once in the final dataframe.


```python
# Keep one complaint per order before merging so the final dataframe remains order-level.
# Without this step, orders with multiple complaints duplicate rows and inflate row counts.
complaints_unique = complaints.drop_duplicates(subset='order_id', keep='first')

# Merge operational context around each order.
df = orders.merge(deliveries, on='order_id', how='left') \
           .merge(customers, on='customer_id', how='left') \
           .merge(complaints_unique, on='order_id', how='left', suffixes=('', '_complaint')) \
           .merge(drivers, left_on='driver_id', right_on='driver_id', how='left') \
           .merge(vehicles, left_on='vehicle_id', right_on='vehicle_id', how='left') \
           .merge(hubs, left_on='hub_id', right_on='hub_id', how='left')

print(f"Consolidated dataframe: {df.shape[0]} rows, {df.shape[1]} columns")
```

Step 4: Verify Data Integrity

```python
# Verify data integrity after merge â€” row counts, duplicates, and coverage
print(f"Original orders: {len(orders)}")
print(f"Merged dataframe rows: {len(df)}")
print(f"Rows match: {len(orders) == len(df)}")

print(f"\nDuplicate order_ids: {df['order_id'].duplicated().sum()}")

print(f"\nOrders with delivery data: {df['delivery_id'].notna().sum()} (expected: 950)")
print(f"Orders with complaint data: {df['complaint_id'].notna().sum()} (expected: 320)")
print(f"Orders with driver assigned: {df['driver_id'].notna().sum()} (expected: 950)")
print(f"Orders with vehicle assigned: {df['vehicle_id'].notna().sum()} (expected: 950)")
print(f"Orders with hub assigned: {df['hub_id'].notna().sum()} (expected: 950)")
```

Step 5: Create Derived Columns.

```python
# Create derived columns for delivery outcome and duration
df['is_failed'] = df['delivery_status'] == 'Failed'
df['is_delayed'] = df['delivery_status'] == 'Delayed'
df['is_ontime'] = df['delivery_status'] == 'OnTime'

# Convert timestamps and calculate delivery duration
df['dispatch_time'] = pd.to_datetime(df['dispatch_time'])
df['delivery_completed_at'] = pd.to_datetime(df['delivery_completed_at'])
df['delivery_duration_hours'] = (df['delivery_completed_at'] - df['dispatch_time']).dt.total_seconds() / 3600

# Summary
print(f"Average delivery duration: {df['delivery_duration_hours'].mean():.2f} hours")
print(f"Failed: {df['is_failed'].sum()} | Delayed: {df['is_delayed'].sum()} | On-time: {df['is_ontime'].sum()}")
```

Exporting cleaned dataset for Mongodb


```python
# Export cleaned dataframes as CSV files in the notebook runtime.
# These filenames match the files stored in cleaned_dataset/ and later loaded by the MongoDB notebook.
customers.to_csv("customers_cleaned.csv", index=False)
orders.to_csv("orders_cleaned.csv", index=False)
deliveries.to_csv("deliveries_cleaned.csv", index=False)
drivers.to_csv("drivers_cleaned.csv", index=False)
vehicles.to_csv("vehicles_cleaned.csv", index=False)
hubs.to_csv("hubs_cleaned.csv", index=False)
complaints.to_csv("complaints_cleaned.csv", index=False)
incidents.to_csv("incidents_cleaned.csv", index=False)
app_events.to_csv("app_events_cleaned.csv", index=False)
data_dictionary.to_csv("data_dictionary_cleaned.csv", index=False)

print("All 10 cleaned files exported")
```

IV. Statistical Analysis with Pandas and NumPy

### Statistical Analysis Objective

This section uses the cleaned and integrated dataframe to quantify operational patterns, including delivery cost behaviour, failure rates by zone, and relationships between numeric performance indicators.

1. Delivery Cost Summary Statistics

```python
# Summary statistics for delivery costs
cost_stats = df['fuel_or_charge_cost'].describe()
print(cost_stats)
```

2. Delivery Failure Rate by Zone (NumPy)

```python
# Failure rate by zone using NumPy
zones = df['zone'].dropna().unique()
for zone in sorted(zones):
    zone_data = df[df['zone'] == zone]
    failure_rate = np.mean(zone_data['is_failed']) * 100
    print(f"{zone}: {failure_rate:.2f}%")
```

3. Correlation Analysis of Numerical Features

```python
# Select numerical columns and calculate correlation matrix
num_cols = ['order_value', 'route_distance_km', 'fuel_or_charge_cost',
            'manual_route_override_count', 'customer_rating_post_delivery']
corr_matrix = df[num_cols].corr().round(3)
print(corr_matrix)
```

V. Data Visualisation

### Visualisation Objective

These charts convert the cleaned analysis outputs into visual evidence for the NorthStar case study, especially around delivery outcomes, complaint types, and relationships between operational variables.

1. Correlation Heatmap

```python
#Correlation heatmap of numerical features
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.3f',
            square=True, linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features', fontsize=14)
plt.tight_layout()
plt.show()
```

3. Delivery Outcome Distribution by Zone

```python
# Delivery Outcome Distribution by Zone
zone_outcome = df.groupby('zone')[['is_failed', 'is_delayed', 'is_ontime']].mean() * 100
zone_outcome = zone_outcome.sort_values('is_failed', ascending=False)

zone_outcome.plot(kind='bar', figsize=(10, 6), color=['red', 'orange', 'green'])
plt.title('Delivery Outcomes by Zone (%)', fontsize=14)
plt.xlabel('Zone')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(['Failed', 'Delayed', 'On-Time'])
plt.tight_layout()
plt.show()
```
